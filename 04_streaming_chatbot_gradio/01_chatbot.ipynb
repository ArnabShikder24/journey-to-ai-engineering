```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
load_dotenv()  # Load environment variables from .env file
print(os.getenv("GROQ_API_KEY") is not None)
print(os.getenv("GOOGLE_API_KEY") is not None)
```

```python
openai = OpenAI(api_key=os.getenv("GROQ_API_KEY"), base_url="https://api.groq.com/openai/v1")
gemini = OpenAI(api_key=os.getenv("GOOGLE_API_KEY"), base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
```

```python
MODEL = "openai/gpt-oss-20b"

system_prompt = "You are a helpful assistant"
```

```python
def chat(message, history):
  history = [{"role":h["role"], "content":h["content"]} for h in history]
  message = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
  stream = openai.chat.completions.create(
    model=MODEL,
    messages=message,
    stream=True
  )
  response = ""
  for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    yield response
```

```python
gr.ChatInterface(
    fn=chat,
    type="messages"
).launch()
```
