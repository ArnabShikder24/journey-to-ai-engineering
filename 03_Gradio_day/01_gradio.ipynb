Gradio is a Python library that allows you to quickly create user interfaces for machine learning models. It provides an easy way to build web-based applications where users can interact with your models through a simple and intuitive interface. With Gradio, you can create input fields, buttons, sliders, and other UI components to collect user input and display model outputs in real-time. This makes it easier to showcase your models, gather feedback, and improve their performance based on user interactions.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
load_dotenv()
print(os.getenv("GROQ_API_KEY") is not None)
print(os.getenv("GOOGLE_API_KEY") is not None)
```

```python
openai = OpenAI(api_key=os.getenv("GROQ_API_KEY"), base_url="https://api.groq.com/openai/v1")
gemini = OpenAI(api_key=os.getenv("GOOGLE_API_KEY"), base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
```

```python
system_prompt = "You are a helpful assistant"

def generate_response_gpt(prompt):
  message = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": prompt}
  ]
  response = openai.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=message
  )
  return response.choices[0].message.content
```

```python
generate_response_gpt("What is today's date?")
```

User Interface

```python
def shout(text):
  print(f"Shout has been called with input : {text}")
  return text.upper()
```

```python
shout("hello world")
```

```python
gr.Interface(fn=shout, inputs="textbox", outputs="textbox", flagging_mode="never").launch()
```

launch(share=True) will give you a public link to your app that you can share with others.
launch(inbrowser=True, auth=("username", "password")) will open the app in your browser and require a username and password to access it.

```python
message_input = gr.Textbox(label="Your message", info="Enter a message to be shouted", lines=7, placeholder="Type your message here...")
message_output = gr.Textbox(label="Shouted message:", info="The shouted message will appear here", lines=8, placeholder="The shouted message will appear here...")

view = gr.Interface(
  fn=shout,
  title="Shout",
  inputs=[message_input],
  outputs=[message_output],
  examples=["hello", "world"],
  flagging_mode="never",
)
view.launch()
```

```python
message_input = gr.Textbox(label="Your prompt", info="Enter a prompt for GPT-OSS-20B", lines=7, placeholder="Type your prompt here...")
message_output = gr.Textbox(label="Response: ",lines=8,)

view = gr.Interface(
  fn=generate_response_gpt,
  title="GPT-OSS-20B",
  inputs=[message_input],
  outputs=[message_output],
  examples=["What is today's date?", "What is the capital of France?"],
  flagging_mode="never",
)
view.launch()
```

```python
system_prompt = "You are a helpful assistant that response in markdown without code blocks"

message_input = gr.Textbox(label="Your prompt", info="Enter a prompt for GPT-OSS-20B", lines=7, placeholder="Type your prompt here...")
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
  fn=generate_response_gpt,
  title="Your Helpful Assistant",
  inputs=[message_input],
  outputs=[message_output],
  examples=[
    "Explain the Transformer architecture to a layperson",
    "Explain the Transformer architecture to an aspiring AI engineer",
  ],
  flagging_mode="never",
)
view.launch()
```

Stream

```python
def stream_gpt(prompt):
  message = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": prompt}
  ]
  response = openai.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=message,
    stream=True
  )
  result = ""
  for chunk in response:
    result += chunk.choices[0].delta.content or ""
    yield result
```

```python
message_input = gr.Textbox(label="Your prompt", info="Enter a prompt for GPT-OSS-20B", lines=7, placeholder="Type your prompt here...")
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
  fn=stream_gpt,
  title="Your Helpful Assistant",
  inputs=[message_input],
  outputs=[message_output],
  examples=[
    "Explain the Transformer architecture to a layperson",
    "Explain the Transformer architecture to an aspiring AI engineer",
  ],
  flagging_mode="never",
)
view.launch()
```

```python
def stream_gemini(prompt):
  message = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": prompt}
  ]
  response = gemini.chat.completions.create(
    model="gemini-3.1-flash-lite",
    messages=message,
    stream=True
  )
  result = ""
  for chunk in response:
    result += chunk.choices[0].delta.content or ""
    yield result
```

```python
def stream_model(prompt, model):
  if model == "GPT":
    result = stream_gpt(prompt)
  elif model == "Gemini":
    result = stream_gemini(prompt)
  else:
    raise ValueError("Invalid model specified")
  yield from result
```

```python
message_input = gr.Textbox(label="Your prompt", info="Enter a prompt for the LLM", lines=7, placeholder="Type your prompt here...")
model_selector = gr.Dropdown(label="Select Model", choices=["GPT", "Gemini"], value="GPT")
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
  fn=stream_model,
  title="Your Helpful Assistant",
  inputs=[message_input, model_selector],
  outputs=[message_output],
  examples=[
    ["Explain the Transformer architecture to a layperson", "GPT"],
    ["Explain the Transformer architecture to an aspiring AI engineer", "Gemini"],
  ],
  flagging_mode="never",
)
view.launch()
```
